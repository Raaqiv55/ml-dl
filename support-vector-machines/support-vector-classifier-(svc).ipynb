```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples=1000, n_features=2, n_redundant=0, n_clusters_per_class=1, n_classes=2)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=10)
```

```python
from sklearn.svm import SVC
```

```python
svc = SVC(kernel="linear")
```

```python
svc.fit(X_train, y_train)
```

```python
y_pred = svc.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(y_test, y_pred))
```

```python
print(confusion_matrix(y_test, y_pred))
```

```python
X, y = make_classification(n_samples=1000, n_features=2, n_redundant=0, n_clusters_per_class=2, n_classes=2)
```

```python
df = pd.DataFrame(X, columns=['Feature_1', 'Feature_2'])


sns.scatterplot(data=df, x='Feature_1', y='Feature_2', hue=y)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=10)
```

```python
svc = SVC(kernel="linear")
svc.fit(X_train, y_train)
```

```python
y_pred1 = svc.predict(X_test)
```

```python
print(classification_report(y_test, y_pred1))
print(confusion_matrix(y_test, y_pred1))
```

```python
rbf = SVC(kernel="rbf")
```

```python
rbf.fit(X_train, y_train)
```

```python
y_pred2 = rbf.predict(X_test)
```

```python
print(classification_report(y_test, y_pred2))
print(confusion_matrix(y_test, y_pred2))
```

```python
polynomial = SVC(kernel="poly")
polynomial.fit(X_train, y_train)

y_pred3 = polynomial.predict(X_test)
print(classification_report(y_test, y_pred3))
print(confusion_matrix(y_test, y_pred3))
```

```python
sigmoid = SVC(kernel="sigmoid")
sigmoid.fit(X_train, y_train)

y_pred4 = sigmoid.predict(X_test)
print(classification_report(y_test, y_pred4))
print(confusion_matrix(y_test, y_pred4))
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {'C': [0.1, 1, 10, 100], 'gamma': [1, 0.1, 0.01, 0.001], 'kernel': ['rbf']}
```

```python
grid = GridSearchCV(SVC(), param_grid = param_grid, refit=True, verbose=2, cv=5)
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
y_pred5 = grid.predict(X_test)
print(classification_report(y_test, y_pred5))
print(confusion_matrix(y_test, y_pred5))
```
