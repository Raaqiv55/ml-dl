```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
```

```python
X, y= make_classification(n_samples=1000,n_features=10,n_classes=2,random_state=42)
```

```python
pd.DataFrame(X)
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
logistic = LogisticRegression()
```

```python
logistic.fit(X_train, y_train)
```

```python
y_pred = logistic.predict(X_test)
print(y_pred) 
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
score = accuracy_score(y_test, y_pred)
print(score)
```

```python
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
print(classification_report(y_test,y_pred))
```

```python
model = LogisticRegression()
penalty = ['l2', 'l1','elasticnet']
c_values = [0.01, 0.1, 1, 10, 100]
solvers = ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga']
```

```python
parmams = dict(penalty=penalty, C=c_values, solver=solvers)
```

```python
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold()
```

```python
from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(estimator=model, param_grid=parmams, scoring='accuracy', cv=cv, n_jobs=-1)
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_test, y_pred)
print(score)
print(classification_report(y_pred,y_test))
print(confusion_matrix(y_pred, y_test))
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
model = LogisticRegression()
randomcv = RandomizedSearchCV(estimator=model, param_distributions=parmams, scoring='accuracy', cv=5, n_jobs=-1)
```

```python
randomcv.fit(X_train, y_train)
```

```python
randomcv.best_params_
```

```python
randomcv.best_score_
```

```python
y_pred = randomcv.predict(X_test)
```

```python
score = accuracy_score(y_test, y_pred)
print(score)
print(classification_report(y_pred,y_test))
print(confusion_matrix(y_pred, y_test))
```

```python
from sklearn.metrics import roc_curve
from sklearn.metrics import roc_auc_score
```

```python
X, y = make_classification(n_samples=1000, n_classes=2, random_state=1)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)  
```

```python
dummy_model_prob = [0 for _ in range(len(y_test))]
dummy_model_prob
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
model_prob = model.predict_proba(X_test)
```

```python
model_prob
```

```python
model_prob = model_prob[:,1]
```

```python
dummy_model_auc = roc_auc_score(y_test, dummy_model_prob)
model_auc = roc_auc_score(y_test, model_prob)
print(dummy_model_auc)
print(model_auc)
```

```python
dummy_fpr, dummy_fc, _ = roc_curve(y_test, dummy_model_prob)
model_fpr, model_tpr, thresholds = roc_curve(y_test, model_prob)
```

```python
dummy_fpr, model_fpr
```

```python
thresholds
```

```python
plt.plot(dummy_fpr, dummy_fc, linestyle='--',label='Dummy Model')
plt.plot(model_fpr, model_tpr, marker='.',label='Logistic Regression')

plt.xlabel('False Positive Rate')
plt.ylabel("True Positive Rate")

plt.legend()
plt.show()
```

```python
fig = plt.figure(figsize=(20,50))
plt.plot(dummy_fpr, dummy_fc, linestyle='--',label='Dummy Model')
plt.plot(model_fpr, model_tpr, marker='.',label='Logistic Regression')
ax = fig.add_subplot(111)
for xyz in zip(model_fpr, model_tpr, thresholds):
    ax.annotate( '%s' % np.round(xyz[2], 2), xy = (xyz[0], xyz[1]))
    
plt.xlabel('False Positive Rate')
plt.ylabel("True Positive Rate")

plt.show()
```
